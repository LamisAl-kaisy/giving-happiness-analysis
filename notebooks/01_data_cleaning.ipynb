```python
# Data Cleaning

# This notebook prepares the PSID 2017 and 2019 data for analysis by loading the extracted dataset, renaming variables, handling missing values, creating donation measures, and exporting a cleaned dataset.
```

```python
## Load Data
import pandas as pd
```

```python
df = pd.read_excel("../data/J364915_clean.xlsx")
```

```python
rename_map = {
    # 2017
    "ER66001": "release_2017",
    "ER66002": "family_id_2017",
    "ER66017": "age_2017",
    "ER66018": "sex_2017",
    "ER66025": "life_satisfaction_2017",
    "ER66164": "employment_status_2017",
    "ER68420": "health_status_2017",
    "ER71040": "donor_status_2017",
    "ER71042": "religious_donations_2017",
    "ER71044": "combo_donations_2017",
    "ER71046": "needy_donations_2017",
    "ER71048": "health_donations_2017",
    "ER71050": "education_donations_2017",
    "ER71052": "youth_donations_2017",
    "ER71054": "cultural_donations_2017",
    "ER71056": "community_donations_2017",
    "ER71058": "environment_donations_2017",
    "ER71060": "international_peace_donations_2017",
    "ER71063": "other_donations_2017",
    "ER71426": "family_income_2017",
    "ER71485": "wealth_2017",
    "ER71538": "education_2017",
    "ER71540": "marital_status_2017",

    # 2019
    "ER72001": "release_2019",
    "ER72002": "family_id_2019",
    "ER72017": "age_2019",
    "ER72018": "sex_2019",
    "ER72025": "life_satisfaction_2019",
    "ER72164": "employment_status_2019",
    "ER74428": "health_status_2019",
    "ER77062": "donor_status_2019",
    "ER77064": "religious_donations_2019",
    "ER77066": "combo_donations_2019",
    "ER77068": "needy_donations_2019",
    "ER77070": "health_donations_2019",
    "ER77072": "education_donations_2019",
    "ER77074": "youth_donations_2019",
    "ER77076": "cultural_donations_2019",
    "ER77078": "community_donations_2019",
    "ER77080": "environment_donations_2019",
    "ER77082": "international_peace_donations_2019",
    "ER77085": "other_donations_2019",
    "ER77448": "family_income_2019",
    "ER77511": "wealth_2019",
    "ER77599": "education_2019",
    "ER77601": "marital_status_2019"
}

df = df.rename(columns=rename_map)
```

```python
df.columns.tolist()
```

```python
check_cols = [
    "life_satisfaction_2017",
    "life_satisfaction_2019",
    "health_status_2017",
    "health_status_2019",
    "donor_status_2017",
    "donor_status_2019",
    "sex_2017",
    "sex_2019",
    "employment_status_2017",
    "employment_status_2019",
    "marital_status_2017",
    "marital_status_2019"
]

for col in check_cols:
    print("\n", col)
    print(df[col].value_counts(dropna=False).sort_index())
```

```python
import numpy as np

# Replace PSID special missing-value codes with NaN

for col in ["age_2017", "age_2019"]:
    df[col] = df[col].replace(999, np.nan)

for col in ["life_satisfaction_2017", "life_satisfaction_2019"]:
    df[col] = df[col].replace([0, 8, 9], np.nan)

for col in ["health_status_2017", "health_status_2019"]:
    df[col] = df[col].replace([8, 9], np.nan)

for col in ["employment_status_2017", "employment_status_2019"]:
    df[col] = df[col].replace(99, np.nan)

for col in ["education_2017", "education_2019"]:
    df[col] = df[col].replace(99, np.nan)

for col in ["marital_status_2017", "marital_status_2019"]:
    df[col] = df[col].replace(9, np.nan)
```

```python
df["donor_2017"] = df["donor_status_2017"].map({
    1: 1,
    5: 0
})

df["donor_2019"] = df["donor_status_2019"].map({
    1: 1,
    5: 0
})
```

```python
donation_cols_2017 = [
    "religious_donations_2017",
    "combo_donations_2017",
    "needy_donations_2017",
    "health_donations_2017",
    "education_donations_2017",
    "youth_donations_2017",
    "cultural_donations_2017",
    "community_donations_2017",
    "environment_donations_2017",
    "international_peace_donations_2017",
    "other_donations_2017"
]

donation_cols_2019 = [
    "religious_donations_2019",
    "combo_donations_2019",
    "needy_donations_2019",
    "health_donations_2019",
    "education_donations_2019",
    "youth_donations_2019",
    "cultural_donations_2019",
    "community_donations_2019",
    "environment_donations_2019",
    "international_peace_donations_2019",
    "other_donations_2019"
]

special_donation_codes = [99998, 99999, 999998, 999999]

df[donation_cols_2017] = df[donation_cols_2017].replace(
    special_donation_codes, np.nan
)

df[donation_cols_2019] = df[donation_cols_2019].replace(
    special_donation_codes, np.nan
)
```

```python
df["total_donations_2017"] = df[donation_cols_2017].sum(axis=1, min_count=1)

df["total_donations_2019"] = df[donation_cols_2019].sum(axis=1, min_count=1)
```

```python
df[
    [
        "donor_2017",
        "total_donations_2017",
        "donor_2019",
        "total_donations_2019"
    ]
].head(10)
```

```python
for year in [2017, 2019]:
    donor = f"donor_{year}"
    total = f"total_donations_{year}"

    print(f"\n{year}")
    print("Non-donor but donations > 0:",
          ((df[donor] == 0) & (df[total] > 0)).sum())

    print("Donor but donations = 0:",
          ((df[donor] == 1) & (df[total] == 0)).sum())
```

```python
df["giving_rate_2017"] = np.where(
    df["family_income_2017"] > 0,
    (df["total_donations_2017"] / df["family_income_2017"]) * 100,
    np.nan
)

df["giving_rate_2019"] = np.where(
    df["family_income_2019"] > 0,
    (df["total_donations_2019"] / df["family_income_2019"]) * 100,
    np.nan
)
```

```python
df[
    [
        "family_income_2017",
        "total_donations_2017",
        "giving_rate_2017",
        "family_income_2019",
        "total_donations_2019",
        "giving_rate_2019"
    ]
].head(10)
```

```python
df.loc[
    (df["donor_2019"] == 0) &
    (df["total_donations_2019"] > 0),
    [
        "family_id_2019",
        "donor_status_2019",
        "total_donations_2019"
    ] + donation_cols_2019
]
```

```python
df.loc[
    (df["donor_2017"] == 1) &
    (df["total_donations_2017"] == 0),
    "total_donations_2017"
] = np.nan

df.loc[
    (df["donor_2019"] == 1) &
    (df["total_donations_2019"] == 0),
    "total_donations_2019"
] = np.nan
```

```python
df["giving_rate_2017"] = np.where(
    df["family_income_2017"] > 0,
    (df["total_donations_2017"] / df["family_income_2017"]) * 100,
    np.nan
)

df["giving_rate_2019"] = np.where(
    df["family_income_2019"] > 0,
    (df["total_donations_2019"] / df["family_income_2019"]) * 100,
    np.nan
)
```

```python
print(
    "2017 donor but zero:",
    ((df["donor_2017"] == 1) & (df["total_donations_2017"] == 0)).sum()
)

print(
    "2019 donor but zero:",
    ((df["donor_2019"] == 1) & (df["total_donations_2019"] == 0)).sum()
)
```

```python
# Flag the contradictory 2019 record
df["donor_inconsistency_2019"] = (
    (df["donor_2019"] == 0) &
    (df["total_donations_2019"] > 0)
)

# Do not force this contradictory case into donor/non-donor analysis
df.loc[
    df["donor_inconsistency_2019"],
    "donor_2019"
] = np.nan
```

```python
print("2019 contradictory donor records:",
      df["donor_inconsistency_2019"].sum())
```
